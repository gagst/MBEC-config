```python
SAVE_FIGS = True

HEATMAP_CELL_SIZE = 0.6
HEATMAP_WIDTH_PADDING = 4.5
HEATMAP_HEIGHT_PADDING = 1.5
HEATMAP_YLABEL_X = 0.04
HEATMAP_ROW_GROUPS = (1, 5, 5, 2)
HEATMAP_GROUP_GAP = 0.18
```

```python
import seaborn as sns
import matplotlib
import matplotlib.pyplot as plt

matplotlib.rcParams['pdf.fonttype'] = 42
matplotlib.rcParams['ps.fonttype'] = 42

def heatmap_figsize(data, cell_size=HEATMAP_CELL_SIZE):
    return (
        HEATMAP_WIDTH_PADDING + data.shape[1] * cell_size,
        HEATMAP_HEIGHT_PADDING + len(data) * cell_size,
    )


def split_heatmap_row_groups(data):
    groups = []
    start = 0
    for group_size in HEATMAP_ROW_GROUPS:
        if start >= len(data):
            break
        end = min(start + group_size, len(data))
        groups.append(data.iloc[start:end])
        start = end
    if start < len(data):
        groups.append(data.iloc[start:])
    return groups


def heatmap_with_row_groups(data, xlabel=None, ylabel=None, title=None, **kwargs):
    groups = split_heatmap_row_groups(data)
    heatmap_kwargs = kwargs.copy()
    heatmap_kwargs.setdefault("vmin", data.min().min())
    heatmap_kwargs.setdefault("vmax", data.max().max())
    fig = plt.gcf()
    axes = fig.subplots(
        len(groups),
        1,
        sharex=True,
        gridspec_kw={"height_ratios": [len(group) for group in groups], "hspace": HEATMAP_GROUP_GAP},
    )
    if len(groups) == 1:
        axes = [axes]

    for index, (ax, group) in enumerate(zip(axes, groups)):
        sns.heatmap(group, ax=ax, **heatmap_kwargs)
        ax.set_ylabel("")
        ax.set_yticklabels(ax.get_yticklabels(), rotation=0)
        ax.tick_params(axis="y", pad=2)
        if index < len(axes) - 1:
            ax.set_xlabel("")
            ax.tick_params(axis="x", labelbottom=False, bottom=False)
        elif xlabel:
            ax.set_xlabel(xlabel)

    if ylabel:
        fig.supylabel(ylabel, x=HEATMAP_YLABEL_X, y=0.5, fontsize=axes[-1].xaxis.label.get_size())
    if title:
        axes[0].set_title(title, loc="right")
    plt.sca(axes[-1])
    return axes
```

```python
from configurator.enums.evaluation import EquipmentEval, ComponentEval, SlotComponentEval, SlotEquipmentEval


def count_equipment_slots(
    df,
    group_by_columns: list,
    no_pivot=False,
    what_to_count="correct_share",
    configurator_order=CONFIGURATOR_NAME_ORDER,
):
    result_types = [member.value for member in SlotEquipmentEval]

    if what_to_count not in ["correct_share", "total_number", "old_total_number"] + [
        member.value for member in SlotEquipmentEval
    ]:
        raise ValueError(
            f"Invalid value for what_to_count: {what_to_count}. Must be 'correct_share', 'total_number' or one of the result types."
        )

    extended_group_columns = group_by_columns.copy()
    extended_group_columns.insert(0, "configurator")
    extended_group_columns.append(
        "result"
    )  # result is the column with the evaluation result found in the input dataframe
    result_columns = extended_group_columns.copy()
    result_columns.remove("result")
    result_columns.append(what_to_count)

    # Count number of each result type for each configurator and groupb_by columns
    counts = df.groupby(extended_group_columns).size().unstack(fill_value=0)

    # Add result types if they were not present in the data
    for rtype in result_types:
        if rtype not in counts.columns:
            counts[rtype] = 0

    # Count total number of slots
    counts["old_total_number"] = counts[result_types].sum(axis=1)
    counts["total_number"] = counts[
        [
            SlotEquipmentEval.CORRECT_VALUE.value,
            SlotEquipmentEval.INCORRECT_VALUE.value,
            SlotEquipmentEval.NO_MATCH_SLOT.value,
            SlotEquipmentEval.VECTOR_VALUE_NO_MATCH.value,
        ]
    ].sum(axis=1)

    # Extract configurator and group_by columns from index
    counts = counts.reset_index()

    # Calculate share of the correctly identified component slots
    counts["correct_share"] = (counts[SlotEquipmentEval.CORRECT_VALUE.value] / counts["total_number"]).round(2)

    # Leave only share of correctly identified component slots
    counts = counts[result_columns]

    if no_pivot:
        counts = counts.set_index("configurator").reindex(configurator_order).reset_index()
        return counts

    # Pivot the table to have configurators as rows and group_by values as columns, with correct_share as values
    counts = counts.pivot(index="configurator", columns=group_by_columns, values=what_to_count)
    counts = counts.reindex(index=configurator_order)
    return counts


def count_component_slots(
    df,
    group_by_columns: list,
    no_pivot=False,
    what_to_count="correct_share",
    configurator_order=CONFIGURATOR_NAME_ORDER,
):
    result_types = [member.value for member in SlotComponentEval]

    if what_to_count not in ["correct_share", "total_number", "old_total_number"] + [
        member.value for member in SlotComponentEval
    ]:
        raise ValueError(
            f"Invalid value for what_to_count: {what_to_count}. Must be 'correct_share', 'total_number' or one of the result types."
        )

    extended_group_columns = group_by_columns.copy()
    extended_group_columns.insert(0, "configurator")
    extended_group_columns.append(
        "result"
    )  # result is the column with the evaluation result found in the input dataframe
    result_columns = extended_group_columns.copy()
    result_columns.remove("result")
    result_columns.append(what_to_count)

    # Count number of each result type for each configurator and groupb_by columns
    counts = df.groupby(extended_group_columns).size().unstack(fill_value=0)

    # Add result types if they were not present in the data
    for rtype in result_types:
        if rtype not in counts.columns:
            counts[rtype] = 0

    # Count total number of slots
    counts["old_total_number"] = counts[result_types].sum(axis=1)
    counts["total_number"] = counts[
        [
            SlotComponentEval.CORRECT_VALUE.value,
            SlotComponentEval.INCORRECT_VALUE.value,
            SlotComponentEval.NO_MATCH_SLOT.value,
            SlotComponentEval.VECTOR_VALUE_NO_MATCH.value,
        ]
    ].sum(axis=1)

    # Extract configurator and group_by columns from index
    counts = counts.reset_index()

    # Calculate share of the correctly identified component slots
    counts["correct_share"] = (counts[SlotComponentEval.CORRECT_VALUE.value] / counts["total_number"]).round(2)

    # Leave only share of correctly identified component slots
    counts = counts[result_columns]

    if no_pivot:
        counts = counts.set_index("configurator").reindex(configurator_order).reset_index()
        return counts

    # Pivot the table to have configurators as rows and group_by values as columns, with correct_share as values
    counts = counts.pivot(index="configurator", columns=group_by_columns, values=what_to_count)
    counts = counts.reindex(index=configurator_order)
    return counts


def count_equipment(
    df,
    group_by_columns: list,
    no_pivot=False,
    what_to_count="fully_correct_share",
    configurator_order=CONFIGURATOR_NAME_ORDER,
):
    result_types = [member.value for member in EquipmentEval]

    if what_to_count not in ["match_share", "fully_correct_share", "total_number"] + result_types:
        raise ValueError(
            f"Invalid value for what_to_count: {what_to_count}. Must be 'match_share', 'fully_correct_share', 'total_number' or one of the result types."
        )

    extended_group_columns = group_by_columns.copy()
    extended_group_columns.insert(0, "configurator")
    extended_group_columns.append(
        "result"
    )  # result is the column with the evaluation result found in the input dataframe
    result_columns = extended_group_columns.copy()
    result_columns.remove("result")
    result_columns.append(what_to_count)

    # Count number of each result type for each configurator and groupb_by columns
    counts = df.groupby(extended_group_columns).size().unstack(fill_value=0)

    # Add result types if they were not present in the data
    for rtype in result_types:
        if rtype not in counts.columns:
            counts[rtype] = 0

    # Count total number of equipment
    counts["total_number"] = counts[result_types].sum(axis=1)

    # Extract configurator and group_by columns from index
    counts = counts.reset_index()

    # Calculate share of the correctly identified equipment
    counts["match_share"] = (1 - counts[EquipmentEval.NO_MATCH.value] / counts["total_number"]).round(2)
    counts["fully_correct_share"] = (counts[EquipmentEval.FULLY_CORRECT.value] / counts["total_number"]).round(2)

    # Leave only share of correctly identified equipment
    counts = counts[result_columns]

    if no_pivot:
        counts = counts.set_index("configurator").reindex(configurator_order).reset_index()
        return counts

    # Pivot the table to have configurators as rows and group_by values as columns, with correct_share as values
    counts = counts.pivot(index="configurator", columns=group_by_columns, values=what_to_count)
    counts = counts.reindex(index=configurator_order)
    return counts


def count_components(
    df,
    group_by_columns: list,
    no_pivot=False,
    what_to_count="fully_correct_share",
    configurator_order=CONFIGURATOR_NAME_ORDER,
):
    result_types = [member.value for member in ComponentEval]

    if what_to_count not in ["match_share", "fully_correct_share", "total_number"] + result_types:
        raise ValueError(
            f"Invalid value for what_to_count: {what_to_count}. Must be 'match_share', 'fully_correct_share', 'total_number' or one of the result types."
        )

    extended_group_columns = group_by_columns.copy()
    extended_group_columns.insert(0, "configurator")
    extended_group_columns.append(
        "result"
    )  # result is the column with the evaluation result found in the input dataframe
    result_columns = extended_group_columns.copy()
    result_columns.remove("result")
    result_columns.append(what_to_count)

    # Count number of each result type for each configurator and groupb_by columns
    counts = df.groupby(extended_group_columns).size().unstack(fill_value=0)

    # Add result types if they were not present in the data
    for rtype in result_types:
        if rtype not in counts.columns:
            counts[rtype] = 0

    # Count total number of components
    counts["total_number"] = counts[result_types].sum(axis=1)

    # Extract configurator and group_by columns from index
    counts = counts.reset_index()

    # Calculate share of the correctly identified components
    counts["match_share"] = (1 - counts[ComponentEval.NO_MATCH.value] / counts["total_number"]).round(2)
    counts["fully_correct_share"] = (counts[ComponentEval.FULLY_CORRECT.value] / counts["total_number"]).round(2)

    # Leave only share of correctly identified components
    counts = counts[result_columns]

    if no_pivot:
        counts = counts.set_index("configurator").reindex(configurator_order).reset_index()
        return counts

    # Pivot the table to have configurators as rows and group_by values as columns, with correct_share as values
    counts = counts.pivot(index="configurator", columns=group_by_columns, values=what_to_count)
    counts = counts.reindex(index=configurator_order)
    return counts
```

```python
def split_model_source(df, name):
    if name == "index":
        values = df.index.to_series()
    else:
        values = df[name]
    # Split at the first '|' character
    split = values.str.split("|", n=1, expand=True)
    # LLM: before |, Source: after | (strip whitespace)
    llm = split[0].str.strip()
    source = split[1].str.strip() if split.shape[1] > 1 else ""
    # Insert as first columns
    df.insert(0, "LLM", llm)
    df.insert(1, "Source", source)
    df.reset_index(drop=True, inplace=True)
    df.set_index("LLM", inplace=True)
    return df


slots = count_equipment_slots(combined_df, [], no_pivot=True, configurator_order=CONFIGURATOR_NAME_ORDER_BIG)
slots = (slots.set_index("configurator") * 100).astype(int)

equipment = count_equipment(combined_df, [], no_pivot=True, configurator_order=CONFIGURATOR_NAME_ORDER_BIG)
equipment = (equipment.set_index("configurator") * 100).astype(int)

result = {}
for configurator, (without_ex, with_ex) in WITHOUT_EXAMPLES_VS_WITH_EXAMPLES.items():
    result[configurator] = {
        "equipment_without_ex": equipment.loc[without_ex, "fully_correct_share"],
        "equipment_with_ex": equipment.loc[with_ex, "fully_correct_share"],
        "equipment_improvement": (
            equipment.loc[with_ex, "fully_correct_share"] - equipment.loc[without_ex, "fully_correct_share"]
        ),
        "slots_without_ex": slots.loc[without_ex, "correct_share"],
        "slots_with_ex": slots.loc[with_ex, "correct_share"],
        "slots_improvement": (slots.loc[with_ex, "correct_share"] - slots.loc[without_ex, "correct_share"]),
    }

result_df = pd.DataFrame.from_dict(result, orient="index")
result_df = split_model_source(result_df, "index")
print(result_df.to_latex())
```
